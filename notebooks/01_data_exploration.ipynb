# Data Exploration

Load preprocessed experiment metadata for UCI, REES46, and ASOS datasets and display summary statistics.

```python
import numpy as np
import os
import sys

# Ensure repo root is on the path
REPO_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.insert(0, REPO_ROOT)
```

## UCI Online Retail II

```python
uci_meta = np.load(os.path.join(REPO_ROOT, 'data/preprocessed/uci/experiments_metadata.npy'),
                   allow_pickle=True)

print(f'Number of UCI experiments: {len(uci_meta)}')
print()
for i, exp in enumerate(uci_meta):
    print(f"  Exp {i}: D0={exp['D0']}, D1={exp['D1']}, "
          f"N_pilot={exp['N_pilot']:,}, U_true={exp['U_true']:,}")
```

## REES46 Rolling Windows (k=21)

```python
rees46_k21 = np.load(os.path.join(REPO_ROOT, 'data/preprocessed/rees46/experiments_rolling_k21.npy'),
                     allow_pickle=True)

print(f'Number of REES46 rolling k=21 experiments: {len(rees46_k21)}')
print()
for i, exp in enumerate(rees46_k21[:5]):
    print(f"  Exp {i}: D0={exp['D0']}, D1={exp['D1']}, "
          f"N_pilot={exp['N_pilot']:,}, U_true={exp['U_true']:,}")
if len(rees46_k21) > 5:
    print(f'  ... ({len(rees46_k21) - 5} more experiments)')
```

## Cumulative User Curve (UCI Experiment 6)

```python
import matplotlib.pyplot as plt

exp = uci_meta[6]
cum = np.array(exp['cumulative_users'])
D0 = exp['D0']

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(range(len(cum)), cum, 'k-', linewidth=1.5)
ax.axvline(x=D0, color='gray', linestyle=':', alpha=0.7, label=f'Pilot cutoff (D0={D0})')
ax.axvspan(0, D0, alpha=0.08, color='gray')
ax.set_xlabel('Day')
ax.set_ylabel('Cumulative distinct users')
ax.set_title(f'UCI Experiment 6: N_pilot={exp["N_pilot"]:,}, U_true={exp["U_true"]:,}')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
