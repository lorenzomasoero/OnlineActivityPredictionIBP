# Accuracy Comparison

Load pre-fitted results for all datasets and compute a summary accuracy table (median v per method per dataset).

```python
import numpy as np
import os
import sys

REPO_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.insert(0, REPO_ROOT)
```

## Load results

```python
datasets = {
    'UCI': np.load(os.path.join(REPO_ROOT, 'results/uci/uci_all_results.npy'),
                   allow_pickle=True).item(),
    'REES46 (k=21)': np.load(os.path.join(REPO_ROOT, 'results/rees46/rees46_rolling_k21_results.npy'),
                              allow_pickle=True).item(),
    'REES46 (k=50)': np.load(os.path.join(REPO_ROOT, 'results/rees46/rees46_rolling_k50_results.npy'),
                              allow_pickle=True).item(),
    'REES46 (k=100)': np.load(os.path.join(REPO_ROOT, 'results/rees46/rees46_rolling_k100_results.npy'),
                               allow_pickle=True).item(),
    'ASOS': np.load(os.path.join(REPO_ROOT, 'results/asos/asos_all_results.npy'),
                    allow_pickle=True).item(),
}

methods = ['TG_SSP', 'IBP', 'NB_SSP_regression']
method_labels = {'TG_SSP': 'TG-SSP', 'IBP': 'IBP', 'NB_SSP_regression': 'NB-SSP (reg)'}

print(f"{'Dataset':<20} {'N_exp':>6}")
print('-' * 28)
for name, res in datasets.items():
    print(f"{name:<20} {len(res):>6}")
```

## Median accuracy (v) per method per dataset

```python
# Header
header = f"{'Dataset':<20}" + ''.join(f'{method_labels[m]:>15}' for m in methods)
print(header)
print('-' * len(header))

for ds_name, results in datasets.items():
    row = f'{ds_name:<20}'
    for m in methods:
        accuracies = []
        for exp_id, r in results.items():
            if m in r and isinstance(r[m], dict) and 'accuracy_v' in r[m]:
                accuracies.append(r[m]['accuracy_v'])
        if accuracies:
            row += f'{np.median(accuracies):>15.3f}'
        else:
            row += f'{"N/A":>15}'
    print(row)
```
