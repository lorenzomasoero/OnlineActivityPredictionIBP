# Hitting-Time MAE

Load pre-computed hitting-time MAE results and display summary tables. The hitting time measures how many additional days a method predicts are needed to reach a target number of new users.

```python
import pandas as pd
import os
import sys

REPO_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.insert(0, REPO_ROOT)
```

## REES46 Rolling k=21

```python
df_k21 = pd.read_csv(os.path.join(REPO_ROOT, 'results/hitting_time/hitting_time_mae_k21.csv'))
print('Hitting-time MAE — REES46 rolling k=21')
print(df_k21.to_string(index=False))
```

## UCI

```python
df_uci = pd.read_csv(os.path.join(REPO_ROOT, 'results/hitting_time/hitting_time_mae_uci.csv'))
print('Hitting-time MAE — UCI')
print(df_uci.to_string(index=False))
```

## Summary: MAE at eta=2.0 across datasets

```python
ht_files = {
    'REES46 k=21': 'hitting_time_mae_k21.csv',
    'REES46 k=50': 'hitting_time_mae_k50.csv',
    'REES46 k=100': 'hitting_time_mae_k100.csv',
    'UCI': 'hitting_time_mae_uci.csv',
    'ASOS': 'hitting_time_mae_asos.csv',
}

print(f"{'Dataset':<18} {'Method':<18} {'MAE':>8} {'Median AE':>10}")
print('-' * 56)

for ds_name, fname in ht_files.items():
    fpath = os.path.join(REPO_ROOT, 'results/hitting_time', fname)
    if not os.path.exists(fpath):
        print(f'{ds_name:<18} (file not found)')
        continue
    df = pd.read_csv(fpath)
    # Filter to eta=2.0
    df2 = df[df['eta'] == 2.0]
    for _, row in df2.iterrows():
        print(f"{ds_name:<18} {row['method']:<18} {row['mae']:>8.2f} {row['median_ae']:>10.1f}")
```

**Interpretation:** Lower MAE indicates more accurate hitting-time predictions. The eta parameter controls the target multiplier — at eta=2.0, we ask: "how many more days until we see 2x the pilot users?" Methods with lower MAE give more reliable answers to this planning question.
