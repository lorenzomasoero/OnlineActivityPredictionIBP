# Paper Figures

Reproduce the key figures from the paper by running the plotting scripts. Figures are saved to `output/` and displayed inline.

```python
import os
import sys
import subprocess
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

REPO_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.insert(0, REPO_ROOT)
OUTPUT_DIR = os.path.join(REPO_ROOT, 'output')
```

## Case Study Trajectories

Prediction trajectories vs. truth for selected experiments across UCI, REES46, and ASOS.

```python
subprocess.run([sys.executable, os.path.join(REPO_ROOT, 'plotting/case_studies.py')],
               cwd=REPO_ROOT, check=True)

for fname in ['fig_case_study_uci.pdf', 'fig_case_study_rees46.pdf', 'fig_case_study_asos.pdf']:
    fpath = os.path.join(OUTPUT_DIR, fname)
    if os.path.exists(fpath):
        print(f'Generated: {fname}')
```

## Hitting-Time Decision Figure

Trajectory panel + error boxplots showing how methods compare for hitting-time prediction.

```python
subprocess.run([sys.executable, os.path.join(REPO_ROOT, 'plotting/decision_figure.py')],
               cwd=REPO_ROOT, check=True)

for fname in os.listdir(OUTPUT_DIR):
    if 'decision' in fname:
        print(f'Generated: {fname}')
```

## Other Figures

Run the remaining plotting scripts. Some may be stubs (marked TODO) and will print a message instead of generating a figure.

```python
scripts = [
    ('Cross-dataset comparison', 'plotting/cross_dataset.py'),
    ('Power-law visualizations', 'plotting/powerlaw_viz.py'),
    ('CI coverage analysis', 'plotting/ci_coverage.py'),
    ('Accuracy figures', 'plotting/accuracy_figures.py'),
]

for label, script in scripts:
    print(f'\n--- {label} ---')
    result = subprocess.run([sys.executable, os.path.join(REPO_ROOT, script)],
                           cwd=REPO_ROOT, capture_output=True, text=True)
    if result.stdout:
        print(result.stdout.strip())
    if result.returncode != 0 and result.stderr:
        print(f'  (error: {result.stderr.strip()[:200]})')
```
